```python
import pandas as pd
import numpy as np

data = {
    "Student Name": ["Aditi", "Rahul", "Priya", "Aman", "Aditi", "Neha"],
    "Age": ["22", "21", "23", "22", "22", "Twenty"],
    "Join Date": ["2026-01-10", "2026-02-15", "2026-03-20", "2026-04-05", "2026-01-10", "2026-05-12"],
    "Marks": [85, 78, np.nan, 91, 85, 88],
    "Department": ["Data Science", "data science", "AI & ML", "AI & ML", "Data Science", None]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python


print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing Marks with the average marks
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Fill missing Department with "Unknown"
df["Department"] = df["Department"].fillna("Unknown")

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nUpdated Dataset:")
display(df)
```

```python
# Convert Age from text to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Convert Join Date from text to datetime
df["Join Date"] = pd.to_datetime(df["Join Date"], errors="coerce")

print("Updated Data Types:")
print(df.dtypes)

print("\nUpdated Dataset:")
display(df)
```

```python
# Handle invalid/missing Age values
df["Age"] = df["Age"].fillna(df["Age"].median())

# Standardize Department names
df["Department"] = df["Department"].str.strip().str.title()

print("Dataset after fixing Age and Department:")
display(df)
```

```python
# Check duplicate rows before removing them
print("Duplicate rows before cleaning:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())

print("\nDataset after removing duplicates:")
display(df)
```

```python
# Standardize column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("&", "and")
)

print("Standardized Column Names:")
print(df.columns.tolist())

print("\nFinal Dataset:")
display(df)
```

```python
# Save the cleaned dataset

output_file = "cleaned_dataset.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", output_file)

# Display the final cleaned dataset
display(df)
```
